```python
# Set the data location and load the precipitation CSV.
DATA_BASE = "https://raw.githubusercontent.com/pjbitterman/KSU_x9076/main/data"
import pandas as pd
precip = pd.read_csv(f"{DATA_BASE}/part1/ohio_precip_2025.csv")
precip.head()
```

```python
# Count the stations and check the precipitation stat
print(f"{len(precip)} stations")
precip["precip_mm_2025"].describe()
```

```python
# Sort precipitation from highest to lowest
wettest = precip.sort_values("precip_mm_2025", ascending=False)  # ascending False for highest to lowest
# Show the names and precipitation values of the five wettest stations
wettest[["name", "precip_mm_2025"]].head()
```

```python
# Calculate mean and maximum precipitation
mean = precip["precip_mm_2025"].mean()
maximum = precip["precip_mm_2025"].max()
# Calculate diffrence to find out how much wettest station above the mean
difference = maximum - mean
print(difference)
```

The wettest station received 486.84 mm more precipitation than the mean. I am not surprised by this difference because Ohio has different geographic conditions, and rainfall can vary from place to place depending on elevation, location, and how storms move across the state.
